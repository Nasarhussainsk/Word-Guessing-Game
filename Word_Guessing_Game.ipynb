## Word Guessing Game

## 🎮 How to Play

1. Start the game by running the Jupyter Notebook.
2. The game randomly selects a word from the word list.
3. The selected word is hidden from the player.
4. Enter **one letter at a time** to guess the word.
5. You have a maximum of **6 incorrect attempts**.
6. **Only incorrect guesses reduce your attempts.** Correct guesses do not reduce your attempts.
7. If you enter a letter that you have already guessed, it will **not count as a new attempt**.
8. Continue guessing until you reveal the complete word or use all 6 incorrect attempts.
9. You win when you successfully guess all the letters in the word.
10. You lose when you reach 6 incorrect attempts without guessing the word.

### 🔢 Attempts

* **Maximum attempts:** 6 incorrect guesses
* **Correct guess:** Does not reduce attempts
* **Incorrect guess:** Reduces attempts by 1
* **Repeated guess:** Does not reduce attempts and is not accepted as a new guess

### 💡 Example

```text
Word: _ _ _ _ _ _

Enter a letter: p
Correct!

Enter a letter: p
You already guessed that letter. Try another one.

Enter a letter: z
Incorrect!
Attempts remaining: 5
```

**Keep track of your guesses and try to reveal the word before you run out of attempts! 🎯**


````python
import random

words = ["Python","Machine","Java script","Amsterdam"]
word = random.choice(words).lower()

guessed_letters = []
attempts = 6

while attempts > 0:
    display = ""

    for letter in word:
        if letter in guessed_letters:
            display += letter
        else:
            display += "_"

    print("Word:", display)

    if "_" not in display:
        print("Congratulations! You guessed the word:", word)
        break

    try:
        guess = input("Enter your letter: ").lower()

        if len(guess) != 1 or not guess.isalpha():
            print("Please enter only one alphabet.")
            continue

        if guess in guessed_letters:
            print("You have already guessed that letter.")
            continue

        guessed_letters.append(guess)

        if guess in word:
            print("Correct Guess!")
        else:
            attempts -= 1
            print("Wrong Guess!")
            print("Attempts left:", attempts)

    except Exception:
        print("Invalid Input.")

if attempts == 0:
    print("Game Over!")
    print("The correct word was:", word)
````
